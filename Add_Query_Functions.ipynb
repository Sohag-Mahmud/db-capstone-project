```python
join_query ="""
select concat(Customers.FirstName, ' ', Customers.LastName) as CustomerName,
Customers.Email, Customers.Phone, Orders.BillAmount from Customers inner join Bookings
on Customers.CustomerID = Bookings.CustomerID
inner join Orders on Bookings.BookingID = Orders.BookingID where Orders.BillAmount > 20;
"""
cursor.execute(join_query)
```
